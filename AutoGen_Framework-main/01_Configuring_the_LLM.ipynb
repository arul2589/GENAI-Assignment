# 01. Configuring the LLM (the agent's brain)

**AutoGen** is a framework from Microsoft for building **AI agents** — programs that can think and act for you.

Before we build any agent, we need to give it a **brain**: a Large Language Model (LLM).
In this whole course we use **OpenAI's `gpt-4o-mini`** as the brain.

The object that connects AutoGen to the brain is called a **model client**.

## Real-life analogy

Think of an agent as a **robot**.

- The **body** (arms, wheels) = AutoGen.
- The **brain** = the LLM (gpt-4o-mini).
- The **model client** = the **wire** that connects the body to the brain.

No brain wire = the robot cannot think. So we always set this up first.

## What you need

| Thing | Why |
|-------|-----|
| `autogen-agentchat` | The main AutoGen library (agents & teams) |
| `autogen-ext[openai]` | The OpenAI connector |
| `OPENAI_API_KEY` | Your secret key, kept in a `.env` file |
| `gpt-4o-mini` | The model (brain) we use everywhere |

Install once in a terminal:
```
pip install -U "autogen-agentchat" "autogen-ext[openai]" python-dotenv
```

````python
# Load the secret OPENAI_API_KEY from the .env file in this folder
from dotenv import load_dotenv
load_dotenv()

# Build the model client (the wire to the brain)
from autogen_ext.models.openai import OpenAIChatCompletionClient

model_client = OpenAIChatCompletionClient(model="gpt-4o-mini")

print("Model client is ready:", model_client)
````

## Test the brain directly (optional)

We can send one message straight to the brain to check it answers.
Every agent will use this same client underneath.

````python
from autogen_core.models import UserMessage

# Ask the brain one simple question
reply = await model_client.create(
    [UserMessage(content="Say hello in 3 words.", source="user")]
)

print(reply.content)
````

## Key points to remember

- **AutoGen** builds AI agents; every agent needs a **brain** (an LLM).
- We use **OpenAI `gpt-4o-mini`** as the brain in this whole course.
- The **model client** is the wire: `OpenAIChatCompletionClient(model="gpt-4o-mini")`.
- Keep your **`OPENAI_API_KEY`** in a `.env` file and load it with `load_dotenv()`.
- We will reuse this `model_client` in every notebook.
