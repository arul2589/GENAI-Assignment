# 10. Teams in AutoGen (overview)

One agent is useful. But hard jobs are easier when **several agents work together** —
each with its own role. A group of agents working together is a **team**.

This notebook is a **gentle overview** of teams. The next notebooks build and run them.

## Real-life analogy

Making a video needs a **team**:

- a **writer** (script),
- an **editor** (improves it),
- a **reviewer** (approves it).

Each person has one job. Together they finish faster and better than one person alone.
AutoGen teams work the same way.

## Types of teams in AutoGen

| Team type | How it picks who talks next | Good for |
|-----------|-----------------------------|----------|
| **`RoundRobinGroupChat`** | Takes turns in order (1, 2, 3, 1, 2...) | Simple, predictable flows |
| **`SelectorGroupChat`** | The brain picks the best next agent | Smart, flexible flows |

We will mostly use **`RoundRobinGroupChat`** because it is the **simplest** to understand.

## The 4 ingredients of a team

| Ingredient | What it is |
|------------|-----------|
| **Agents** | The members, each with a role |
| **Team type** | RoundRobin (take turns) or Selector |
| **Termination** | A rule for when to **stop** (notebook 10) |
| **Task** | The job you give the whole team |

## Key points to remember

- A **team** is several agents working together, each with a **role**.
- **`RoundRobinGroupChat`** = agents take turns in order (the simple default).
- **`SelectorGroupChat`** = the brain chooses who speaks next (more advanced).
- Every team needs a **termination** rule so it knows when to **stop**.
