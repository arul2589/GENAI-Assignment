You now know the building blocks. The **official documentation** is where you go
to learn more and look up details. This notebook is a **map** so you know where to look.

## The AutoGen family (what is what)

| Package | What it is | We used it for |
|---------|-----------|----------------|
| **autogen-agentchat** | The friendly, high-level API | Agents & teams (this whole course) |
| **autogen-core** | The low-level engine | Memory, message types |
| **autogen-ext** | Connectors & extras | The OpenAI model client |

For beginners, **autogen-agentchat** is the main one you use.

## How to read docs as a beginner

| Tip | Why |
|-----|-----|
| Start with the **AgentChat tutorial** | It matches what we learned |
| **Copy a code example**, run it | Learning by doing is fastest |
| Change **one thing** at a time | See what each part does |
| Search for the **class name** (e.g. `RoundRobinGroupChat`) | Jump straight to details |
| Ignore the advanced "Core" parts at first | Not needed for beginners |
