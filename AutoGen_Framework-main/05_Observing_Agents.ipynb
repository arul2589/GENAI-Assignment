# 05. Observing Agents (watch them work live)

So far we waited for the **final** answer with `run()`.

But we can also **watch the agent work step by step**, live, as it types.
This is called **streaming**, and AutoGen gives us a simple helper called **`Console`** to print it nicely.

## Real-life analogy

- `run()` is like getting a **finished letter** in the mail — you only see the end result.
- Streaming with `Console` is like **watching someone type** in a live chat — you see it appear word by word, step by step.

Watching live helps you **understand and debug** what the agent is doing.

## Two ways to get answers

| Method | What you see | Good for |
|--------|--------------|----------|
| `await agent.run(task=...)` | Only the final answer | Simple, quick results |
| `await Console(agent.run_stream(task=...))` | Every step, live | Watching & understanding |

Note the small change: **`run_stream`** instead of `run`, wrapped in **`Console(...)`**.

```python
from dotenv import load_dotenv
load_dotenv()

from autogen_agentchat.agents import AssistantAgent
from autogen_agentchat.ui import Console
from autogen_ext.models.openai import OpenAIChatCompletionClient

model_client = OpenAIChatCompletionClient(model="gpt-4o-mini")
agent = AssistantAgent(
    name="storyteller",
    model_client=model_client,
    system_message="You tell very short, fun stories.",
)

# Console prints every step live as the agent works
await Console(agent.run_stream(task="Tell a 3-sentence story about a brave cat."))
```

## Key points to remember

- **Observing** an agent means watching it work **live**, step by step.
- This is called **streaming**.
- Use **`run_stream(...)`** instead of `run(...)`, and wrap it in **`Console(...)`**.
- `Console` prints each step nicely so you can **understand and debug** the agent.
- `run()` = only the final answer; `Console(run_stream())` = the whole journey.
