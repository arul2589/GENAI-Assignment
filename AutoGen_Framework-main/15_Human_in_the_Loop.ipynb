# 16. Human in the Loop

Sometimes the team should **pause and ask YOU** before continuing.
This is called **human in the loop** — a real person joins the conversation.

AutoGen has a special agent for this: the **`UserProxyAgent`**.
When it is that agent's turn, **you** get to type the reply.

## Real-life analogy

It is like a **team that checks with the boss**:

- The agents do the work,
- but before finishing they turn to **you** and ask, "Is this okay?"
- You type your answer, and the team continues based on it.

## How it works

| Piece | Job |
|-------|-----|
| `AssistantAgent` | The AI member(s) |
| `UserProxyAgent` | **You** — it asks you to type input |
| `RoundRobinGroupChat` | Takes turns: AI, then you, then AI... |
| Termination | e.g. stop when you type "DONE" |

> When you run the cell, a **text box appears** — type your reply and press Enter.

```python
from dotenv import load_dotenv
load_dotenv()

from autogen_agentchat.agents import AssistantAgent, UserProxyAgent
from autogen_agentchat.teams import RoundRobinGroupChat
from autogen_agentchat.conditions import TextMentionTermination
from autogen_agentchat.ui import Console
from autogen_ext.models.openai import OpenAIChatCompletionClient

model_client = OpenAIChatCompletionClient(model="gpt-4o-mini")

# The AI assistant
assistant = AssistantAgent(
    name="assistant",
    model_client=model_client,
    system_message="You help plan a small party. Ask the user short questions and suggest ideas.",
)

# YOU. When it is this agent's turn, a box appears for you to type.
user = UserProxyAgent(name="user")

# Stop when YOU type the word DONE
team = RoundRobinGroupChat([assistant, user], termination_condition=TextMentionTermination("DONE"))

# Try it: answer the assistant's questions. Type DONE when you are happy.
await Console(team.run_stream(task="Help me plan a birthday party."))
```

## Key points to remember

- **Human in the loop** = a real person joins the agent conversation.
- Use the **`UserProxyAgent`** — when it is its turn, **you** type the reply.
- Put it in a team with AI agents so they **take turns** with you.
- A typing box appears when it is your turn (type and press Enter).
- Use a termination like `TextMentionTermination("DONE")` so **you** can end it.
