# 04. Messages

Everything in AutoGen is made of **messages**.

- You send a **task** (a message) to the agent.
- The agent replies with **messages**.
- A whole conversation is just a **list of messages**.

Each message knows **who said it** (`source`) and **what was said** (`content`).

## Real-life analogy

Messages are like **chat bubbles** in WhatsApp.

- Each bubble has a **sender** (you or your friend) and **text**.
- The full chat is the **list of bubbles** from top to bottom.

AutoGen works the same way.

## The message types you will see

| Type | Who sends it | Example |
|------|--------------|---------|
| `TextMessage` | You (the user) | "What is 2 + 2?" |
| `TextMessage` | The agent | "It is 4." |
| (later) tool messages | The agent's tools | results from a tool |

For now, just remember: a message has **`.source`** (who) and **`.content`** (what).

```python
from dotenv import load_dotenv
load_dotenv()

from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient

model_client = OpenAIChatCompletionClient(model="gpt-4o-mini")
agent = AssistantAgent(name="assistant", model_client=model_client)

result = await agent.run(task="What are the 3 primary colours?")

# result.messages is the LIST of all chat bubbles
for m in result.messages:
    print(f"[{m.source}] -> {m.content}")
```

## Building a message yourself

You can also create a `TextMessage` by hand. This is useful later when you control the chat.

```python
from autogen_agentchat.messages import TextMessage

# A message I write myself (who = "user", what = the text)
my_message = TextMessage(content="Hello agent!", source="user")

print("Who said it:", my_message.source)
print("What was said:", my_message.content)
```

## Key points to remember

- A conversation is just a **list of messages** (`result.messages`).
- Every message has **`.source`** (who) and **`.content`** (what).
- The user's **task** is a message; the agent's **reply** is a message.
- The agent's final answer is the **last** message: `result.messages[-1].content`.
- The basic message type is **`TextMessage`**.
