# 02. Your First Agent

An **agent** is the brain (LLM) plus a **role** and the ability to **do tasks** for you.

The most common agent in AutoGen is the **`AssistantAgent`** — a helpful assistant you give a task, and it replies.

In this notebook we build the simplest possible agent and run it.

## Real-life analogy

An agent is like a **helpful employee**:

- You hire them and tell them their **job** (the `system_message`).
- You give them a **task** (`run(task=...)`).
- They **reply** with the finished work.

## The 3 simple steps

| Step | Code |
|------|------|
| 1. Brain | `OpenAIChatCompletionClient(model="gpt-4o-mini")` |
| 2. Agent | `AssistantAgent(name, model_client=..., system_message=...)` |
| 3. Run it | `await agent.run(task="...")` |

```python
from dotenv import load_dotenv
load_dotenv()

from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient

# 1) Brain
model_client = OpenAIChatCompletionClient(model="gpt-4o-mini")

# 2) Agent (give it a name and a job)
assistant = AssistantAgent(
    name="assistant",
    model_client=model_client,
    system_message="You are a friendly helper. Keep answers short.",
)

# 3) Run it. In a notebook we use "await".
result = await assistant.run(task="Give me one fun fact about the moon.")

# The agent's answer is the LAST message
print(result.messages[-1].content)
```

## What did we get back?

`result.messages` is the **list of all messages** in the conversation.
The **last** one is the agent's answer. We will look at messages more closely in notebook 04.

```python
print("Number of messages:", len(result.messages))
print("Who answered:", result.messages[-1].source)
print("Answer:", result.messages[-1].content)
```

## Key points to remember

- The **`AssistantAgent`** is the basic AutoGen agent.
- Give it a **`name`**, a **`model_client`** (the brain), and a **`system_message`** (its job).
- Run it with **`await agent.run(task="...")`** (use `await` inside a notebook).
- The answer is **`result.messages[-1].content`** (the last message).
