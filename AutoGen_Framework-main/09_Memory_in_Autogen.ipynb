# 09. Memory in AutoGen

By default, an agent **forgets** everything after each `run()`.
**Memory** lets an agent **remember facts** and use them in later answers.

AutoGen has a simple memory called **`ListMemory`** — a list of remembered notes
that the agent automatically reads before answering.

## Real-life analogy

Memory is like a **sticky note** on the agent's desk.

- You write "The user's name is Thiru" on a sticky note.
- Every time the agent answers, it **glances at the note** first.
- So it always remembers your name.

## How memory works

| Step | Code |
|------|------|
| 1. Make memory | `memory = ListMemory()` |
| 2. Add a fact | `await memory.add(MemoryContent(content="...", mime_type="text/plain"))` |
| 3. Give it to agent | `AssistantAgent(..., memory=[memory])` |
| 4. Ask | the agent uses the remembered facts |

```python
from dotenv import load_dotenv
load_dotenv()

from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient
from autogen_core.memory import ListMemory, MemoryContent

model_client = OpenAIChatCompletionClient(model="gpt-4o-mini")

# 1) Make a memory and 2) add some facts to remember
memory = ListMemory()
await memory.add(MemoryContent(content="The user's name is Thiru.", mime_type="text/plain"))
await memory.add(MemoryContent(content="The user loves cricket.", mime_type="text/plain"))

# 3) Give the memory to the agent
agent = AssistantAgent(
    name="assistant",
    model_client=model_client,
    memory=[memory],                    # the agent will read these notes
    system_message="You are a friendly assistant.",
)

# 4) Ask something that needs the remembered facts
result = await agent.run(task="What is my name, and suggest a sport I would enjoy.")
print(result.messages[-1].content)
```

## Key points to remember

- By default an agent **forgets** after each run; **memory** makes it remember.
- **`ListMemory`** is a simple list of remembered notes (like sticky notes).
- Add facts with **`await memory.add(MemoryContent(content=..., mime_type="text/plain"))`**.
- Give it to the agent with **`memory=[memory]`**.
- The agent **automatically reads** the notes before answering.
