# 07. Custom Tools with Python

In notebook 06 we used a tiny `add` tool.
Now let's build our **own useful tools** — any Python function can become a tool.

The secret to a good tool: a **clear name**, **typed inputs**, and a **short docstring**
that tells the agent *what the tool does*. The agent reads the docstring to decide when to use it.

## Real-life analogy

Tools are like apps on your phone:

- a **weather app**, a **currency converter**, a **timer**...

Each app does **one clear job**. We will write small Python functions that each do one clear job,
and hand them to the agent as tools.

## Rules for a good custom tool

| Rule | Why |
|------|-----|
| Clear name (`convert_to_celsius`) | The agent guesses use from the name |
| Typed inputs (`temp: float`) | The agent sends the right kind of value |
| Short docstring | Explains what the tool does |
| Return a simple value | Easy for the agent to use |

```python
from dotenv import load_dotenv
load_dotenv()

from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient

model_client = OpenAIChatCompletionClient(model="gpt-4o-mini")

# Custom tool 1: convert Fahrenheit to Celsius
def to_celsius(fahrenheit: float) -> float:
    """Convert a temperature from Fahrenheit to Celsius."""
    return round((fahrenheit - 32) * 5 / 9, 1)

# Custom tool 2: how many days until a given number of weeks
def weeks_to_days(weeks: int) -> int:
    """Return how many days are in the given number of weeks."""
    return weeks * 7

helper = AssistantAgent(
    name="helper",
    model_client=model_client,
    tools=[to_celsius, weeks_to_days],     # give BOTH tools
    reflect_on_tool_use=True,
    system_message="You are a helpful assistant. Use your tools when needed.",
)

result = await helper.run(task="What is 98.6 Fahrenheit in Celsius? Also, how many days are in 3 weeks?")
print(result.messages[-1].content)
```

## Key points to remember

- **Any Python function** can become a custom tool.
- Give it a **clear name**, **typed inputs**, and a **short docstring**.
- You can give an agent **many tools** at once: `tools=[tool1, tool2]`.
- The agent reads each docstring to decide **which** tool to use and **when**.
- Custom tools are how you give your agent **real-world powers**.
