# 06. Tools in Agents

An LLM brain is smart, but it **cannot do everything** by itself.
For example, it is not reliable at exact math, and it cannot look things up.

**Tools** fix this. A tool is just a **Python function** we give to the agent.
The agent can **call the tool** when it needs to, then use the result.

## Real-life analogy

A person is smart, but to add big numbers fast they use a **calculator**.

- The **person** = the LLM brain.
- The **calculator** = a tool.

The agent decides *when* to pick up the tool, uses it, and tells you the answer.

## How tools work

| Step | What happens |
|------|--------------|
| 1 | You give the agent a Python function as a **tool** |
| 2 | You ask a question that needs the tool |
| 3 | The agent **calls the tool** with the right inputs |
| 4 | The agent reads the tool's result and **answers you** |

We set **`reflect_on_tool_use=True`** so the agent turns the tool result into a friendly final sentence.

```python
from dotenv import load_dotenv
load_dotenv()

from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient

model_client = OpenAIChatCompletionClient(model="gpt-4o-mini")

# A tool is just a normal Python function with a short docstring.
def add(a: int, b: int) -> int:
    """Add two numbers and return the result."""
    return a + b

# Give the tool to the agent with tools=[...]
math_agent = AssistantAgent(
    name="math_agent",
    model_client=model_client,
    tools=[add],                     # <-- the tool
    reflect_on_tool_use=True,        # turn the tool result into a nice sentence
    system_message="You are a math helper. Use your tools to calculate.",
)

result = await math_agent.run(task="What is 124 plus 376?")
print(result.messages[-1].content)
```

## Key points to remember

- A **tool** lets an agent **do** things the brain alone cannot (math, lookups, actions).
- A tool is just a **Python function** with a clear name and a short docstring.
- Give tools to an agent with **`tools=[my_function]`**.
- Set **`reflect_on_tool_use=True`** so the agent gives a friendly final answer.
- The agent decides **when** to use a tool, all by itself.
