# 03. Prompt Engineering

**Prompt engineering** means writing **clear instructions** so the agent gives you the answer you want.

In AutoGen there are two places to write instructions:
1. The **`system_message`** — the agent's permanent job/personality.
2. The **`task`** — the one thing you ask right now.

Better instructions = better answers.

## Real-life analogy

Talking to an agent is like giving directions to a **new helper**:

- "Do something" → confusing, you get random results.
- "Write a 2-line birthday wish for my mom, in a warm tone" → clear, you get exactly what you want.

The clearer you are, the better the result.

## Tips for good prompts

| Tip | Weak prompt | Strong prompt |
|-----|-------------|---------------|
| Be specific | "Write about dogs" | "Write 3 lines about why dogs are loyal" |
| Give a role | (none) | "You are a friendly teacher" |
| Say the format | "List fruits" | "List 5 fruits as bullet points" |
| Set the tone | "Reply" | "Reply in a polite, simple tone" |

```python
from dotenv import load_dotenv
load_dotenv()

from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient

model_client = OpenAIChatCompletionClient(model="gpt-4o-mini")

# The system_message sets the ROLE and STYLE (its permanent job)
teacher = AssistantAgent(
    name="teacher",
    model_client=model_client,
    system_message="You are a kind teacher for 10-year-old kids. Explain simply and use a short example.",
)

# A clear, specific task
result = await teacher.run(task="Explain what electricity is in 3 short sentences.")
print(result.messages[-1].content)
```

## Key points to remember

- **Prompt engineering** = writing clear instructions to get better answers.
- Use **`system_message`** for the agent's permanent role and style.
- Use **`task`** for the specific thing you want now.
- Be **specific**: say the topic, the format, the length, and the tone.
- A clear prompt almost always beats a vague one.
