# Pydantic (Data Validation Made Easy)

**Pydantic** is a Python library that **checks your data** and makes sure it is the **correct type** and shape.

You describe what your data should look like (a *model*), and Pydantic makes sure any data you give it follows those rules.

## Real-life analogy

Think of Pydantic like a **security guard at an entrance**.

- You set the rule: *"only people with a valid ticket may enter."*
- The guard **checks every person** at the door.
- If someone has no ticket, the guard **stops them** and tells you why.

Pydantic is that guard for your **data**.

## Why use Pydantic?

| Without Pydantic | With Pydantic |
|------------------|---------------|
| You manually check every value | Checking is automatic |
| Wrong data slips through | Wrong data is caught early |
| `"25"` (text) vs `25` (number) confusion | It converts and checks types for you |
| Long, messy `if` checks | Clean, short model classes |

## Step 1: Make a model

A model is just a class that lists the **fields** and their **types**.

```python
# Install once (in a terminal):  pip install pydantic
from pydantic import BaseModel

# This model says: a User must have a name (text) and an age (number)
class User(BaseModel):
    name: str
    age: int

print("Model created!")
```

## Step 2: Create an object from data

Pydantic checks the data as soon as you create the object.

```python
user = User(name="Thiru", age=25)

print(user)
print("Name:", user.name)
print("Age:", user.age)
```

## Step 3: Automatic type conversion

If the value *can* be turned into the right type, Pydantic does it for you.

```python
# Age is given as TEXT "30", but the model wants a number
user2 = User(name="Asha", age="30")

print(user2)
print("Age is now a real number:", user2.age + 5)   # works! 30 + 5 = 35
```

## Step 4: Catching bad data

If the data breaks the rules, Pydantic raises a clear error telling you what's wrong.

```python
from pydantic import ValidationError

try:
    bad = User(name="Ravi", age="hello")   # "hello" is NOT a number
except ValidationError as e:
    print("Pydantic caught the problem:")
    print(e)
```

## Step 5: Default values and optional fields

You can give a field a default value so it is not always required.

```python
from typing import Optional

class Product(BaseModel):
    name: str
    price: float
    in_stock: bool = True          # default value
    discount: Optional[float] = None  # optional, can be left out

p = Product(name="Pen", price=10.5)
print(p)
```

## Step 6: Models inside models (nested)

A field can be **another model**, so you can build bigger structured data.

```python
class Address(BaseModel):
    city: str
    pincode: str

class Person(BaseModel):
    name: str
    address: Address   # a model inside a model

person = Person(
    name="Meera",
    address={"city": "Chennai", "pincode": "600001"},
)

print(person)
print("City:", person.address.city)
```

## Key points to remember

- **Pydantic** checks that your data has the **correct type and shape**.
- A **model** is a class built from `BaseModel` that lists fields and their types.
- It **automatically converts** values when it safely can (e.g. `"30"` → `30`).
- Bad data raises a clear **`ValidationError`** so you catch mistakes early.
- Use **default values** and `Optional` for fields that are not always required.
- Models can be **nested** to describe bigger, structured data.
- Pydantic is used a lot in **FastAPI, LangChain, and CrewAI** to keep data safe.
